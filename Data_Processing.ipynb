```python
# Task 9: Python Data Processing

# Function to validate marks
def validate_marks(marks):
    return (
        isinstance(marks, (int, float))
        and not isinstance(marks, bool)
        and 0 <= marks <= 100
    )


# Function to calculate summary statistics
def calculate_statistics(marks):
    if not marks:
        return None

    return {
        "total": sum(marks),
        "average": sum(marks) / len(marks),
        "highest": max(marks),
        "lowest": min(marks)
    }


# Student records using a list of dictionaries
students = [
    {"name": "Ramya", "marks": 90},
    {"name": "Vanita", "marks": 85},
    {"name": "Jagdish", "marks": 78},
    {"name": "Chetan", "marks": 92},
    {"name": "Lilly", "marks": 88}
]


# Validate and process student records
valid_students = []

for student in students:
    if validate_marks(student["marks"]):
        valid_students.append(student)
    else:
        print(f"Invalid marks for {student['name']}.")


# Extract marks into a list
marks_list = [student["marks"] for student in valid_students]

# Calculate statistics
statistics = calculate_statistics(marks_list)


# Display student data processing report
print("=" * 50)
print("       STUDENT DATA PROCESSING REPORT")
print("=" * 50)

print("\nStudent Records")
print("-" * 50)
print(f"{'Student Name':<20}{'Marks':>10}{'Result':>15}")
print("-" * 50)

for student in valid_students:
    result = "PASS" if student["marks"] >= 35 else "FAIL"
    print(
        f"{student['name']:<20}"
        f"{student['marks']:>10}"
        f"{result:>15}"
    )


# Display summary statistics
if statistics is not None:
    top_student = max(
        valid_students,
        key=lambda student: student["marks"]
    )

    lowest_student = min(
        valid_students,
        key=lambda student: student["marks"]
    )

    passed_students = [
        student for student in valid_students
        if student["marks"] >= 35
    ]

    failed_students = [
        student for student in valid_students
        if student["marks"] < 35
    ]

    print("\nSummary Statistics")
    print("-" * 50)
    print(f"Total Students  : {len(valid_students)}")
    print(f"Total Marks     : {statistics['total']}")
    print(f"Average Marks   : {statistics['average']:.2f}")
    print(
        f"Highest Marks   : {statistics['highest']} "
        f"({top_student['name']})"
    )
    print(
        f"Lowest Marks    : {statistics['lowest']} "
        f"({lowest_student['name']})"
    )
    print(f"Students Passed : {len(passed_students)}")
    print(f"Students Failed : {len(failed_students)}")

    print("\nData processing completed successfully!")

else:
    print("\nNo valid student records available.")

print("=" * 50)
```
